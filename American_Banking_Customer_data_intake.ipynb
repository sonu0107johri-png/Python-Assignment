```python
year_counts = {}

for chunk in pd.read_csv(file_path, usecols=["Date received"], chunksize=500000):
    years = pd.to_datetime(chunk["Date received"]).dt.year.value_counts()
    for year, count in years.items():
        year_counts[year] = year_counts.get(year, 0) + count

year_counts
```

```python
total_complaints = sum(year_counts.values())

total_complaints
```

```python
dict(sorted(year_counts.items()))
```

```python
output_path = "../data/Cleaned/complaints_2025.csv"

first_chunk = True

for chunk in pd.read_csv(file_path, chunksize=500000):
    chunk["Date received"] = pd.to_datetime(chunk["Date received"], errors="coerce")
    complaints_2025 = chunk[chunk["Date received"].dt.year == 2025]
    complaints_2025.to_csv(output_path, mode="w" if first_chunk else "a", header=first_chunk, index=False)
    first_chunk = False
```
